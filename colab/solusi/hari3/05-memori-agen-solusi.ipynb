## Bab 5 - Solusi Latihan 1: Asisten Serah Terima Shift dengan Memori Persisten
Jalankan sel satu per satu dengan Shift+Enter

Asisten yang mengingat setiap operator, bahkan setelah program dimulai ulang.
- Memori disimpan di file SQLite (SqliteSaver)
- Setiap nama operator memiliki thread sendiri
- Perintah pada mode interaktif: /riwayat, /baru, /keluar

Bagian utama berupa demo berskrip yang menirukan: pakai aplikasi, tutup, lalu buka lagi.
Sel terakhir membuka mode interaktif (ketik pesan sendiri) setelah `INTERAKTIF` diubah menjadi `True`.

```python
# Langkah 1 - Impor dan fungsi bantu
import os
import sqlite3
import tempfile
import uuid

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.checkpoint.sqlite import SqliteSaver

from migas_shared.config import ROOT

load_dotenv()

DB_BAWAAN = str(ROOT / "hari3" / "05-memori-agen" / "memori_chat.db")


def buat_agen(db_path: str):
    model = ChatOpenAI(
        model=os.getenv("ONLINE_MODEL"),
        base_url=os.getenv("ONLINE_BASE_URL"),
        api_key=os.getenv("ONLINE_API_KEY"),
        reasoning_effort=os.getenv("LLM_REASONING_EFFORT", "none"),
    )
    connection = sqlite3.connect(db_path, check_same_thread=False)
    agent = create_agent(
        model,
        tools=[],
        system_prompt=(
            "Anda asisten serah terima shift untuk operator produksi migas. Jawab singkat dalam Bahasa Indonesia. "
            "Ingat apa yang dikatakan pengguna tentang dirinya dan pekerjaannya."
        ),
        checkpointer=SqliteSaver(connection),
    )
    return agent, connection


def chat(agent, user: str, sesi: str, teks: str) -> str:
    """Satu giliran percakapan. user dan sesi bersama-sama menentukan thread."""
    config = {"configurable": {"thread_id": f"{user}:{sesi}"}}
    response = agent.invoke({"messages": [HumanMessage(content=teks)]}, config)
    return response["messages"][-1].content


def jumlah_pesan(agent, user: str, sesi: str) -> int:
    config = {"configurable": {"thread_id": f"{user}:{sesi}"}}
    return len(agent.get_state(config).values.get("messages", []))


# Folder sementara untuk file database demo; dibersihkan di sel terakhir demo
folder = tempfile.TemporaryDirectory()
db_path = os.path.join(folder.name, "demo.db")
```

## Langkah 2 - Jalan 1: pakai aplikasi
Rina memperkenalkan diri dan menyebut jam tersibuknya. Percakapan disimpan di file SQLite.

```python
print("Jalan 1")
agent, connection = buat_agen(db_path)
for teks in ["Saya Rina, supervisor shift malam di GS-SR1.", "Jam tersibuk kami adalah pukul 07.00."]:
    print(f"rina: {teks}")
    print(f"Agen: {chat(agent, 'rina', 'utama', teks)}")
print(f"Pesan tersimpan untuk rina: {jumlah_pesan(agent, 'rina', 'utama')}")
connection.close()
```

## Langkah 3 - Jalan 2: program dimulai ulang
Agen dan koneksi baru membaca file yang sama. Riwayat Rina dimuat, sedangkan Agus belum punya riwayat.

```python
print("Jalan 2 (program dimulai ulang)")
agent, connection = buat_agen(db_path)
print(f"Pesan dimuat untuk rina: {jumlah_pesan(agent, 'rina', 'utama')}")
print(f"Pesan dimuat untuk agus: {jumlah_pesan(agent, 'agus', 'utama')}")
pertanyaan = "Di gathering station mana saya bertugas dan kapan jam tersibuk saya?"
print(f"rina: {pertanyaan}")
print(f"Agen: {chat(agent, 'rina', 'utama', pertanyaan)}")
```

## Langkah 4 - Sesi baru untuk Rina
Sesi baru berarti thread baru, jadi agen tidak mengingat percakapan sebelumnya.

```python
print("Sesi baru untuk rina (thread baru)")
print(f"rina: {pertanyaan}")
print(f"Agen: {chat(agent, 'rina', 'sesi2', pertanyaan)}")
connection.close()
folder.cleanup()
```

## Langkah 5 - Mode interaktif (opsional)
Ubah `INTERAKTIF` menjadi `True`, jalankan sel, lalu ketik nama dan pesan Anda. Memori disimpan di
`hari3/05-memori-agen/memori_chat.db`, jadi percakapan berlanjut saat sel dijalankan lagi.
Perintah: `/riwayat` (jumlah pesan), `/baru` (sesi baru), `/keluar` (berhenti).

```python
INTERAKTIF = False  # ubah ke True untuk mencoba sendiri

if INTERAKTIF:
    user = input("Nama Anda: ").strip().lower() or "tamu"
    sesi = "utama"
    agent, connection = buat_agen(DB_BAWAAN)

    ada = jumlah_pesan(agent, user, sesi)
    if ada:
        print(f"Selamat datang kembali, {user}! {ada} pesan tersimpan dimuat.")
    print("Perintah: /riwayat  /baru  /keluar\n")

    while True:
        teks = input("Anda: ").strip()
        if teks == "/keluar":
            break
        if teks == "/riwayat":
            print(f"   {jumlah_pesan(agent, user, sesi)} pesan dalam sesi ini")
        elif teks == "/baru":
            sesi = uuid.uuid4().hex[:6]
            print(f"   Sesi baru dimulai ({sesi}). Sesi lama tetap tersimpan.")
        elif teks:
            print(f"Agen: {chat(agent, user, sesi, teks)}\n")
    connection.close()
```

### Penjelasan Solusi Latihan 5.1: memori yang bertahan setelah program dimulai ulang

Memori disimpan di file SQLite lewat `SqliteSaver`, bukan di RAM. Karena itu agen dan koneksi baru yang membuka file yang sama langsung "mengenal" percakapan lama.

- `TODO(B5.1)`: `sqlite3.connect(..., check_same_thread=False)` lalu `create_agent(..., checkpointer=SqliteSaver(connection))`. Koneksi ikut dikembalikan supaya bisa ditutup.
- `TODO(B5.2)`: `thread_id` dibentuk dari `user:sesi`. Pasangan inilah yang memisahkan Rina dari Agus, dan sesi utama dari sesi baru.
- `TODO(B5.3)`: jumlah pesan dibaca dari `agent.get_state(config)`.

Keluaran memperlihatkan tiga hal. Setelah jalan pertama, Rina punya 4 pesan. Setelah "program dimulai ulang" (agen dan koneksi baru), 4 pesan itu termuat kembali, Agus punya 0, dan agen tetap tahu bahwa Rina bertugas di GS-SR1 dengan jam tersibuk pukul 07.00. Pada sesi baru (thread baru), pertanyaan yang sama dijawab dengan tempat dan jam yang karangan model sendiri. Itu bukan bug: tanpa riwayat, model tetap menjawab. Memori yang salah tempat lebih berbahaya daripada tidak punya memori, jadi pemisahan thread perlu disengaja.

## Bab 5 - Solusi Latihan Bonus: Membandingkan Strategi Memori
Jalankan sel satu per satu dengan Shift+Enter

Menjalankan percakapan berskrip YANG SAMA dengan tiga strategi dan mengukur:
- berapa banyak pesan dan (perkiraan) token yang tersimpan di memori
- apakah agen masih bisa mengingat fakta awal di akhir percakapan

Strategi: simpan semua, pangkas ke N pesan terakhir, ringkas pesan lama.

```python
# Langkah 1 - Menyiapkan model
import os

from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware, before_model
from langchain_core.messages import HumanMessage, RemoveMessage
from langchain_core.messages.utils import count_tokens_approximately
from langchain_openai import ChatOpenAI
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.graph.message import REMOVE_ALL_MESSAGES

load_dotenv()

print("Membandingkan Strategi Memori\n")

model = ChatOpenAI(
    model=os.getenv("ONLINE_MODEL"),
    base_url=os.getenv("ONLINE_BASE_URL"),
    api_key=os.getenv("ONLINE_API_KEY"),
    reasoning_effort=os.getenv("LLM_REASONING_EFFORT", "none"),
)
prompt = "Anda asisten operasi yang membantu. Jawab singkat dalam Bahasa Indonesia."
```

## Langkah 2 - Percakapan uji dan fungsi evaluasi
`evaluasi` memutar percakapan berskrip, mengajukan pertanyaan pengingat di akhir, lalu mengukur isi memori.

```python
SIMPAN_TERAKHIR = 4

PERCAKAPAN = [
    "Nomor Permit to Work yang saya pegang adalah PTW-2026-0714.",
    "Pekerjaan kami adalah penggantian gasket flange pada kompresor K-ML1-201.",
    "Pekan lalu terjadi pemadaman listrik selama 2 jam di GS-ML1.",
    "Generator cadangan menyala otomatis setelah 30 detik.",
    "Kemarin kompresor menangani sekitar 1,2 juta scf gas.",
]
PERTANYAAN_INGAT = "Apa nomor Permit to Work yang saya sebutkan di awal?"
DIHARAPKAN = "PTW-2026-0714"


@before_model
def simpan_terbaru(state, runtime):
    messages = state["messages"]
    if len(messages) <= SIMPAN_TERAKHIR:
        return None
    return {"messages": [RemoveMessage(id=REMOVE_ALL_MESSAGES), *messages[-SIMPAN_TERAKHIR:]]}


def evaluasi(nama: str, agent) -> dict:
    config = {"configurable": {"thread_id": nama}}
    for teks in PERCAKAPAN:
        agent.invoke({"messages": [HumanMessage(content=teks)]}, config)

    jawaban = agent.invoke({"messages": [HumanMessage(content=PERTANYAAN_INGAT)]}, config)
    tersimpan = agent.get_state(config).values["messages"]
    return {
        "strategi": nama,
        "pesan": len(tersimpan),
        "token": count_tokens_approximately(tersimpan),
        "ingat": DIHARAPKAN in jawaban["messages"][-1].content,
        "jawaban": jawaban["messages"][-1].content,
    }
```

## Langkah 3 - Tiga agen dengan strategi berbeda

```python
agen = {
    "simpan semua": create_agent(model, tools=[], system_prompt=prompt, checkpointer=InMemorySaver()),
    f"pangkas ke {SIMPAN_TERAKHIR}": create_agent(
        model, tools=[], system_prompt=prompt, middleware=[simpan_terbaru], checkpointer=InMemorySaver()
    ),
    "ringkas": create_agent(
        model,
        tools=[],
        system_prompt=prompt,
        middleware=[SummarizationMiddleware(model=model, trigger=("messages", 6), keep=("messages", 2))],
        checkpointer=InMemorySaver(),
    ),
}

hasil = []
```

## Langkah 4 - Strategi 1: simpan semua

```python
nama = "simpan semua"
print(f"Menjalankan: {nama}")
hasil.append(evaluasi(nama, agen[nama]))
```

## Langkah 5 - Strategi 2: pangkas ke pesan terakhir

```python
nama = f"pangkas ke {SIMPAN_TERAKHIR}"
print(f"Menjalankan: {nama}")
hasil.append(evaluasi(nama, agen[nama]))
```

## Langkah 6 - Strategi 3: ringkas pesan lama

```python
nama = "ringkas"
print(f"Menjalankan: {nama}")
hasil.append(evaluasi(nama, agen[nama]))
```

## Langkah 7 - Tabel perbandingan

```python
print(f"{'Strategi':<20}{'Pesan':>8}{'~Token':>10}   Ingat nomor PTW?")
print("-" * 60)
for h in hasil:
    tanda = "ya" if h["ingat"] else "tidak"
    print(f"{h['strategi']:<20}{h['pesan']:>8}{h['token']:>10}   {tanda}")
```

## Membaca tabel
- Simpan semua: mengingat, tetapi tumbuh tanpa batas
- Pangkas: kecil, tetapi bisa melupakan fakta awal
- Ringkas: menjaga fakta kunci, tetapi ringkasannya sendiri memakai token.
  Pada percakapan PENDEK ringkasan bisa lebih besar dari aslinya; manfaatnya terasa
  pada percakapan panjang (coba perpanjang daftar `PERCAKAPAN`)

Percakapan berskrip yang sama dijalankan pada tiga agen, lalu diukur: berapa pesan dan token yang tersimpan, dan apakah nomor Permit to Work dari pesan pertama masih diingat.
